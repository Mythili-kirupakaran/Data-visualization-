```python
# ==========================================
# DATA VISUALIZATION
# Retail & Supply Chain Analytics
# ==========================================

from google.colab import files
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------
# Upload Dataset
# ------------------------------------------

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print(df.head())


# ==========================================
# 1. Quantity vs Unit Price - Scatter Plot
# ==========================================

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="quantity",
    y="unit_price"
)

plt.title("Quantity vs Unit Price")
plt.xlabel("Quantity")
plt.ylabel("Unit Price")

plt.show()


# ==========================================
# 2. Unit Price Outlier Detection
# ==========================================

plt.figure(figsize=(8, 5))

sns.boxplot(
    x=df["unit_price"]
)

plt.title("Unit Price Outlier Detection")
plt.xlabel("Unit Price")

plt.show()


# ==========================================
# 3. Correlation Heatmap
# ==========================================

plt.figure(figsize=(12, 8))

corr = df[
    [
        "age",
        "cancellations_count",
        "unit_price",
        "quantity",
        "purchase_frequency"
    ]
].corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("E-Commerce Analytics - Correlation Heatmap")

plt.show()


# ==========================================
# 4. Subscription Status Count
# ==========================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="subscription_status"
)

plt.xticks(rotation=45)

plt.title("Customers by Subscription Status")
plt.xlabel("Subscription Status")
plt.ylabel("Number of Records")

plt.show()


# ==========================================
# 5. Average Unit Price Over Time
# ==========================================

df["order_date"] = pd.to_datetime(
    df["order_date"],
    errors="coerce"
)

daily = (
    df.groupby("order_date")["unit_price"]
    .mean()
)

plt.figure(figsize=(12, 5))

plt.plot(
    daily
)

plt.title("Average Unit Price Over Time")
plt.xlabel("Order Date")
plt.ylabel("Average Unit Price")

plt.show()


# ==========================================
# 6. Country Distribution
# ==========================================

country = df["country"].value_counts()

plt.figure(figsize=(8, 8))

plt.pie(
    country,
    labels=country.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Country Distribution")

plt.show()


# ==========================================
# 7. Average Unit Price by Country
# ==========================================

country_avg = (
    df.groupby("country")["unit_price"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 5))

country_avg.plot(
    kind="bar"
)

plt.title("Top Countries by Average Unit Price")
plt.xlabel("Country")
plt.ylabel("Average Unit Price")

plt.show()


# ==========================================
# 8. Unit Price Distribution
# ==========================================

plt.figure(figsize=(8, 5))

sns.histplot(
    df["unit_price"],
    bins=30,
    kde=True
)

plt.title("Unit Price Distribution")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")

plt.show()


# ==========================================
# 9. Pair Plot
# ==========================================

sns.pairplot(
    df[
        [
            "age",
            "unit_price",
            "quantity",
            "purchase_frequency",
            "cancellations_count"
        ]
    ]
)

plt.show()


# ==========================================
# 10. Bubble Chart
# ==========================================

plt.figure(figsize=(9, 6))

plt.scatter(
    df["quantity"],
    df["unit_price"],
    s=df["purchase_frequency"] * 5,
    alpha=0.5,
    c=df["cancellations_count"],
    cmap="viridis"
)

plt.colorbar(
    label="Cancellation Count"
)

plt.xlabel("Quantity")
plt.ylabel("Unit Price")

plt.title(
    "Bubble Chart: Quantity vs Unit Price"
)

plt.show()


# ==========================================
# 11. Violin Plot
# ==========================================

plt.figure(figsize=(10, 6))

sns.violinplot(
    data=df,
    x="category",
    y="unit_price"
)

plt.xticks(rotation=45)

plt.title("Unit Price Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Unit Price")

plt.show()


# ==========================================
# 12. Box Plot
# ==========================================

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="category",
    y="unit_price"
)

plt.xticks(rotation=45)

plt.title("Unit Price by Product Category")
plt.xlabel("Category")
plt.ylabel("Unit Price")

plt.show()
```
